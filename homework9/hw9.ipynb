```python
# Your code here
def plot(x,y):
    """
    Plots a given x and y function.
    input: x and y function
    output: plot
    """
    plt.figure(figsize=(5,5))
    plt.plot(x,y,color='blue',linestyle='-',marker='.',label='line')
    plt.scatter(x,y,color='red',linestyle='-.',marker='o',label='dot')
    plt.title('x vs y',fontsize=20)
    plt.xlabel('x',fontsize=10)
    plt.ylabel('y',fontsize=10)
    plt.legend()
    plt.grid()
    plt.text(3, 6, 'Center Point', fontsize=12, color='red', ha='right') #add text to graphs to emphasize certain portions/point
    plt.show()

my_x=[1,2,3,4,5]
my_y=[2,4,6,8,10]
plot(my_x,my_y)
```

# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
# Your code here
def plot_fx_gx(f,g,x):
    """
    Function plots a given f(x) and g(x) vs an x domain
    inputs=f,g,x
    outputs=plot
    """
    plt.figure(figsize=(10,8))
    plt.plot(x,f,linestyle='--',color='blue',label='f(x)')
    plt.plot(x,g,linestyle=':',color='red',label='g(x)')
    plt.title("f(x) and g(x) vs x")
    plt.xlabel('x-axis')
    plt.ylabel('functions')
    plt.legend()
    plt.show()

my_x=np.linspace(0,10,100)
my_f=my_x**2
my_g=np.sqrt(my_x)
plot_fx_gx(my_f,my_g,my_x)
```

```python
# Your code here
def new_plot(f,g,dom):
    """
    Function plots a given f(x) and g(x) vs an x domain
    inputs=f,g,x
    outputs=plot
    """
    plt.figure(figsize=(10,8))
    plt.plot(dom,f,linestyle='--',color='blue',label='f(x)')
    plt.plot(dom,g,linestyle=':',color='red',label='g(x)')
    sum=f+g
    plt.plot(dom,sum,linestyle='-.',color='purple',label='sum of functions')
    
    plt.title("f(x) and g(x) vs x")
    plt.xlabel('x-axis')
    plt.ylabel('functions')
    
    plt.axvline(x=7,label='vertical line')
    plt.axhline(y=100,label='horizontal line')
    
    plt.xlim(0,15)
    plt.ylim(0,200)
    plt.legend(loc='upper right',bbox_to_anchor=(1, 1.2))
    plt.show()

my_x=np.linspace(0,50,100)
my_f=my_x**2
my_g=np.sqrt(my_x)
new_plot(my_f,my_g,my_x)
```
